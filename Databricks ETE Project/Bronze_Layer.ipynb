
### Dynamic Capabilities

```python
dbutils.widgets.text("file_name", " ")
```

```python
p_filename = dbutils.widgets.get("file_name")
```

# Data Reading

```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "parquet")\
    .option("cloudFiles.schemaLocation", f"abfss://bronze@databrikcsete.dfs.core.windows.net/checkpoint_{p_filename}")\
    .load(f"abfss://source@databrikcsete.dfs.core.windows.net/{p_filename}")
```

```python
df.writeStream.format("parquet")\
    .outputMode("append")\
    .option("checkpointLocation", f"abfss://bronze@databrikcsete.dfs.core.windows.net/checkpoint_{p_filename}")\
    .option("path", f"abfss://bronze@databrikcsete.dfs.core.windows.net/{p_filename}")\
    .trigger(once=True)\
    .start()
```
